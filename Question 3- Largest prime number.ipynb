The prime factors of 13195 are 7, 7, 9 and 29.

What is the largest prime factor of the number 600851475143?

```python
def prime_or_not(n):
    if n < 2 or n % 1 != 0:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


def check_in_list(list):
    filtered_list = []
    for i in range(len(list)):
        if prime_or_not(list[i]) == True:
            filtered_list.append(list[i])
            i = 1+1
        else:
            i = i+1
    return filtered_list

def make_a_list(n):
    factors = [1]
    half = n/2
    i = 1
    while i <= half:
        inverse_factor = n/i
        if inverse_factor % 1 ==0:
            factors.append(int(inverse_factor))
            #factors.append(i)
            i = i +1
        else:
            i = i+1
    return factors

def first_in_list(list):

    for i in range(len(list)):
        if prime_or_not(list[i]) == True:
            return list[i]
        else:
            i = i+1


def all_in_one(n):
    factors = [n]
    i = 0
    while prime_or_not(factors[i]) == False and  factors[i] % 1 < 1:
        i = i+1
        inverse_factor = n / i 
        factors.append((inverse_factor))


    return int(factors[i])


```

```python
print(make_a_list(318))

print(first_in_list(make_a_list(318)))

print(all_in_one(600851475143))




```
