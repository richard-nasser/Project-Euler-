The sum of the squares of the first ten natural numbers is,

1^2 + 2^2 + ... + 10^2 = 385

The square of the sum of the first ten natural numbers is,

(1 + 2 + ... + 10)^2 = 3026

Hence the difference between the sum of the squares of the first ten natural numbers and the square of the sum is: 3025 - 384 = 2640.

Find the difference between the sum of the squares of the first one hundred natural numbers and the square of the sum.

```python
def sumsqrdiff(limit):
    return (sum(range(1, limit+1))**2 - sum(i**2 for i in range(1, limit+1)))
print(sumsqrdiff(10))
print(sumsqrdiff(100))
```
