Problem 0 states: A number is a perfect square, or a square number, if it is the square of a positive integer.
For example, 25 is a square number because 5 ** 2 = 5 * 5 = 25   ; it is also an odd square.

The first 5 square numbers are: 1 4 9 16 25, and the sum of the odd squares is  35  .

Among the first 828 thousand square numbers, what is the sum of all the odd squares?

```python
import numpy as np 



def sum_of_first_n_squares(n):
    list = [ ]
    end = n

    for i in range(1, end + 1):
        list.append(i**2)
        


    return sum(list)

def sum_of_first_n_odd_squares(n):
    list = np.arange(1, n + 1, 2)
    return sum(list**2)
```

```python
print(sum_of_first_n_squares(2))
```

```python
print(sum_of_first_n_odd_squares(828000))
```
