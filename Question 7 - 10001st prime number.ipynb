By listing the first six prime numbers: 2, 3, 5, 7, 11 and 13, we can see that the 6th prime is 13.

What is the 10,001st prime number?

```python


def prime_or_not(n):
    if n < 2 or n % 1 != 0:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


def nthprime(n):
    i = 3
    j = 1
    while j < n:
        if prime_or_not(i) == True:
            j = j+1
        i = i+2
    return i-2

print(nthprime(6))
print(nthprime(10001))
```
